# 🚗 AutoValue AI
## Indian Used-Car Price Prediction & Market Analysis

**Project presentation notebook**

This notebook is a clean presentation of the project. It is intentionally separate from the full model-training notebook, so the original training history and experiments remain preserved.

## 1. Project Objective

The goal is to build an AI/ML system that can estimate the price of an Indian used car from vehicle information.

The project focuses on a complete workflow:

```text
Vehicle Information
        ↓
Data Preparation
        ↓
Machine Learning Model
        ↓
Predicted Used-Car Price
        ↓
Django Web Application
```

The trained Random Forest model is integrated into the project's Django application.

## 2. Dataset

The training work uses an Indian used-car dataset containing vehicle attributes such as:

- Brand
- Model
- Manufacturing year / vehicle age
- Kilometres driven
- Transmission
- Owner information
- Fuel type
- Asking price

The training notebook also contains extensive data-quality investigation. This presentation intentionally shows only the important decisions rather than all exploratory cells.

## 3. Data Preparation

Before training, the data was inspected for:

- Missing values
- Duplicate records
- Invalid or suspicious records
- Incorrect data types
- Mileage formatting
- Extreme values

For example, mileage values stored as text such as `98,000 km` were converted into numerical values so that the machine-learning model could use them.

## 4. Features and Target

The model learns a relationship between vehicle characteristics and price.

### Input features

```text
Brand
Model
Year
Age
Kilometres Driven
Transmission
Owner
Fuel Type
```

### Target

```text
AskPrice
```

In machine learning terminology:

`X = vehicle features`  →  `y = used-car price`

## 5. Preprocessing

Machine-learning algorithms require numerical input. The project therefore separates the features into categorical and numerical groups.

### Categorical features
- Brand
- Model
- Transmission
- Owner
- Fuel Type

### Numerical features
- Year
- Age
- Kilometres Driven

Categorical features are converted using **One-Hot Encoding**. Numerical features are passed through as numerical values.

The preprocessing is fitted on the training data and then applied to the test data. This prevents test-set information from being used while learning the preprocessing.

## 6. Machine Learning Model

### Random Forest Regressor

Random Forest is an ensemble regression algorithm. It builds multiple decision trees and combines their predictions to produce a numerical price estimate.

The project uses:

- `n_estimators = 100`
- `random_state = 42`
- `n_jobs = -1`

The important pipeline is:

```text
Raw Vehicle Data
       ↓
One-Hot Encoding + Numerical Features
       ↓
Random Forest Regressor
       ↓
Predicted Price
```

## 7. Model Evaluation

The model is evaluated on unseen test data using three regression metrics.

| Metric | Meaning |
|---|---|
| MAE | Average absolute prediction error |
| RMSE | Error metric that gives more weight to larger errors |
| R² | Proportion of target variation explained by the model |

The training notebook reports the baseline Random Forest results as approximately:

- **MAE:** ₹158,052
- **RMSE:** ₹507,958
- **R²:** 0.8883

These values are from the documented baseline evaluation on the test set.

## 8. Understanding the Results

The evaluation numbers should be read together rather than as a single score.

- **MAE** gives an intuitive average absolute error in rupees.
- **RMSE** highlights larger prediction mistakes more strongly.
- **R²** describes how much of the variation in prices is explained by the model.

The training notebook also investigates individual prediction errors and price ranges. Those experiments are kept in the full training notebook rather than repeated here.

## 9. Data-Cleaning Experiment

The training work tested whether removing vehicles with more than 500,000 km would improve the model.

The experiment removed 40 training records, but the measured performance became worse:

- Baseline MAE: approximately ₹158,052
- Cleaned-model MAE: approximately ₹161,325
- Baseline R²: approximately 0.8883
- Cleaned-model R²: approximately 0.8811

This is an important machine-learning lesson: an unusual observation is not automatically a bad observation. Data-cleaning decisions should be tested against evidence rather than made only because a value looks extreme.
